```python
# Free open API query to Open-Meteo for atmospheric stability variables
import requests

# Sample coordinates: Dehradun / Uttarakhand (high cloudburst / flash flood vulnerability zone)
latitude = 30.3165
longitude = 78.0322

url = f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&hourly=temperature_2m,relative_humidity_2m,precipitation,cape,surface_pressure&forecast_days=1"
response = requests.get(url)
data = response.json()
print("Fetched hourly atmospheric variables:", list(data.get('hourly', {}).keys()))
```
